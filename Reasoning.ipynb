```python
#Medical Diagnosis-Bayesian reason
print("MEDICAL DIAGNOSIS")

prior = float(input("Enter disease probability: "))
positive = float(input("Enter positive test probability: "))
negative = float(input("Enter false positive probability: "))

numerator = prior * positive
denominator = numerator + (1 - prior) * negative

result = numerator / denominator

print("Probability of Disease:", result * 100, "%")

if result >= 0.5:
    print("High chance of disease")
else:
    print("Low chance of disease")
```

```python
#Bayesian reasning -Electric Vehicle Battery Fault
print("BATTERY FAULT DETECTION")

prior = 0.05
fault = 0.9
normal = 0.1

numerator = prior * fault
denominator = numerator + (1 - prior) * normal

result = numerator / denominator

print("Fault Probability:", result * 100, "%")

if result >= 0.5:
    print("Battery Fault Detected")
else:
    print("Battery Working Normally")
```

```python
#Bayesian Reasoning – Autonomous Vehicle Obstacle Detection
print("OBSTACLE DETECTION")

prior = 0.1
obstacle = 0.95
clear = 0.05

numerator = prior * obstacle
denominator = numerator + (1 - prior) * clear

result = numerator / denominator

print("Obstacle Probability:", result * 100, "%")

if result >= 0.5:
    print("Obstacle Detected - Apply Brake")
else:
    print("Road is Clear")
```

```python

#Fuzzy – Smart Manufacturing Machine Monitoring
print("MACHINE MONITORING")

vibration = 9

if vibration <= 3:
    risk = 0
elif vibration <= 10:
    risk = (vibration - 3) / 7
else:
    risk = 1

print("Machine Risk:", risk)

if risk >= 0.7:
    print("High Risk - Stop Machine")
elif risk >= 0.3:
    print("Medium Risk - Maintenance Required")
else:
    print("Machine Working Normally")
```

```python
#Fuzzy – Medical Fever Detection
print("FEVER DETECTION")

temp = 39.5

if temp <= 37:
    fever = 0
elif temp <= 40:
    fever = (temp - 37) / 3
else:
    fever = 1

print("Fever Level:", fever)

if fever >= 0.7:
    print("High Fever")
elif fever >= 0.3:
    print("Moderate Fever")
else:
    print("Normal Temperature")
```
